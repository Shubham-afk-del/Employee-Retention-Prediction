- The gap is modest and these rows are a small share of the data, so I drop them. Imputing instead would have to wait until after the train-validation split, so the fill values come from training data only, and with about 95% of the rows still left it would add little.

### **2.2 Identify and handle redundant values within categorical columns (if any)** <font color = red>[3 marks]</font>

Examine the categorical columns to determine if any value or column needs to be treated

```python
# Write a function to display the categorical columns with their unique values and check for redundant values
def show_categories(data):
    for col in data.select_dtypes(include="object").columns:
        print(f"{col} ({data[col].nunique()} values)")
        print(data[col].value_counts().to_string())
        print()

show_categories(df)
```

**Observations:**
- Education Level shows `Bachelorâ€™s Degree` and `Masterâ€™s Degree`. This is a curly apostrophe (’) that went through the wrong text encoding at some point, a common UTF-8 and Windows-1252 mix-up. The categories themselves are fine, only the text is garbled, so I replace `â€™` with a plain apostrophe.
- Some labels differ from the data dictionary. Work-Life Balance uses Fair where the dictionary says Below Average, Job Satisfaction and Employee Recognition have Very High rather than Very Low, Company Reputation has Fair rather than Very Poor, and Overtime is Yes/No rather than a number of hours. The values in the data are consistent, so this looks like a problem with the dictionary, and I keep the values as they are.
- Every other column has a small set of distinct labels with no obvious duplicates.

```python
# Replace the garbled apostrophe in Education Level
df["Education Level"] = df["Education Level"].str.replace("â€™", "'", regex=False)
```

```python
# Check the data
print(df["Education Level"].value_counts().to_string())

# Look for labels that differ only by case or extra spaces
variants = [col for col in df.select_dtypes(include="object").columns
            if df[col].str.strip().str.lower().nunique() != df[col].nunique()]
print("\nColumns with case or spacing variants:", variants)
```

**Observation:** Education Level now reads Bachelor's Degree and Master's Degree with the same counts as before (21,169 and 14,303), and no column has labels that differ only by case or extra spaces.

## **3. Train-Validation Split** <font color = red>[5 marks]</font>

### **3.2 Define feature and target variables** <font color = red>[2 Mark]</font>

```python
# Put all the feature variables in X
X = df.drop(columns=["Attrition"])

# Put the target variable in y
y = df["Attrition"]
```

### **4.4 Perform bivariate analysis** <font color="red">[8 Marks]</font>

Perform bivariate analysis on training data between all the categorical columns and target variable to  analyse how the categorical variables influence the target variable.

```python
# Plot distribution for each categorical column with target variable
cat_cols = X_train.select_dtypes(include="object").columns.tolist()

# Put ordered categories in their natural order (the rest go in alphabetical order)
order = {
    "Work-Life Balance": ["Poor", "Fair", "Good", "Excellent"],
    "Job Satisfaction": ["Low", "Medium", "High", "Very High"],
    "Performance Rating": ["Low", "Below Average", "Average", "High"],
    "Education Level": ["High School", "Associate Degree", "Bachelor's Degree", "Master's Degree", "PhD"],
    "Job Level": ["Entry", "Mid", "Senior"],
    "Company Size": ["Small", "Medium", "Large"],
    "Company Reputation": ["Poor", "Fair", "Good", "Excellent"],
    "Employee Recognition": ["Low", "Medium", "High", "Very High"],
}

train_df = X_train.copy()
train_df["Attrition"] = y_train

fig, axes = plt.subplots(5, 3, figsize=(18, 26))
for ax, col in zip(axes.flat, cat_cols):
    sns.countplot(data=train_df, x=col, hue="Attrition", order=order.get(col, sorted(train_df[col].unique())),
                  hue_order=["Stayed", "Left"], palette={"Stayed": blue, "Left": orange}, ax=ax)
    ax.set_title(col)
    ax.set_xlabel("")
    ax.set_ylabel("Number of employees")
    ax.tick_params(axis="x", rotation=20)
plt.tight_layout()
plt.show()
```

```python
# Share of employees who stayed in each category, and the gap between the highest and lowest category
train_df["Stayed"] = (train_df["Attrition"] == "Stayed").astype(int)

rows = []
for c in cat_cols:
    rate = train_df.groupby(c)["Stayed"].mean() * 100
    rows.append([c, rate.idxmax(), rate.max(), rate.idxmin(), rate.min(), rate.max() - rate.min()])

gaps = pd.DataFrame(rows, columns=["Feature", "Highest", "Stayed % (highest)",
                                   "Lowest", "Stayed % (lowest)", "Gap (points)"])
gaps = gaps.sort_values("Gap (points)", ascending=False).reset_index(drop=True)
gaps.round(1)
```

```python
# Job Satisfaction looked odd in the plots, so check it on its own
(train_df.groupby("Job Satisfaction")["Stayed"].mean() * 100).reindex(order["Job Satisfaction"]).round(1)
```

```python
# Plot the share who stayed for the six features with the biggest gaps
base_rate = train_df["Stayed"].mean() * 100

fig, axes = plt.subplots(2, 3, figsize=(16, 9.5))
for ax, c in zip(axes.flat, gaps["Feature"].head(6)):
    rate = train_df.groupby(c)["Stayed"].mean() * 100
    if c in order:
        rate = rate.reindex(order[c])
    ax.bar(range(len(rate)), rate.values, color=blue, width=0.6)
    ax.axhline(base_rate, color=grey, linestyle="--", linewidth=1, zorder=0.9)
    for i, v in enumerate(rate.values):
        ax.text(i, v - 3, f"{v:.0f}%", ha="center", va="top", color="white", fontweight="bold")
    ax.set_xticks(range(len(rate)), [label.replace(" ", "\n") for label in rate.index])
    ax.set_title(c)
    ax.set_ylim(0, 100)
    ax.set_ylabel("Stayed (%)")
fig.suptitle(f"Share of employees who stayed, by category (dashed line: overall rate of {base_rate:.1f}%)",
             fontsize=14)
plt.tight_layout()
plt.savefig("images/retention-by-category.png", dpi=150, bbox_inches="tight")
plt.show()
```

**Observations:** The share who stayed varies most across:
- Job Level: 36.1% for Entry against 79.3% for Senior, the largest gap of any feature (43.3 points).
- Marital Status: 32.5% of single employees stayed, against 63.8% of married ones.
- Remote Work: 75.4% of remote workers stayed, against 47.1% of everyone else.
- Education Level: 75.0% of PhD holders stayed, while every other level sits at 50 to 51%.
- Work-Life Balance: from 40.0% for Poor up to 64.0% for Excellent.

Company Reputation, Performance Rating, Gender and Job Satisfaction show smaller gaps of 8.4 to 12.8 points, and Overtime 5.6 points. Company Size, Innovation Opportunities, Leadership Opportunities, Employee Recognition and Job Role barely move (3.2 points or less), so they are unlikely to matter much in the model.

Job Satisfaction does not run in order: employees with Very High satisfaction stayed at 46.6%, about the same as Low (46.3%) and well below High (54.7%) and Medium (54.3%). That is hard to explain and may be another sign that the data is synthetic.

### **7.3 Find the optimal cutoff** <font color = red>[15 marks]</font>

Find the optimal cutoff to improve model performance. While a default threshold of 0.5 was used for initial evaluation, optimising this threshold can enhance the model's performance.

First, plot the ROC curve and check AUC.



7.3.1 Plot ROC curve <font color="red">[3 Marks]</font>

```python
# Define ROC function
def draw_roc(actual, probs, title, filename=None):
    fpr, tpr, thresholds = metrics.roc_curve(actual, probs, drop_intermediate=False)
    auc_score = metrics.roc_auc_score(actual, probs)

    plt.figure(figsize=(6, 5.5))
    plt.plot(fpr, tpr, color=blue, linewidth=2, label=f"Logistic regression (AUC = {auc_score:.3f})")
    plt.plot([0, 1], [0, 1], color=grey, linestyle="--", linewidth=1, label="Random guess")
    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.02])
    plt.xlabel("False positive rate (1 - specificity)")
    plt.ylabel("True positive rate (sensitivity)")
    plt.title(title)
    plt.legend(loc="lower right")
    if filename:
        plt.savefig(filename, dpi=150, bbox_inches="tight")
    plt.show()
    return auc_score
```

```python
# Call the ROC function
auc_train = draw_roc(y_train_pred_final["Stayed"], y_train_pred_final["Stayed_Prob"],
                     "ROC curve (training data)", "images/roc-curve.png")
print("Training AUC:", round(auc_train, 4))
```

**Observation:** The training AUC is 0.8257. Put simply, if you pick one employee who stayed and one who left at random, the model gives the stayer the higher probability about 83% of the time. The curve sits well above the diagonal across its whole range.

**Sensitivity and Specificity tradeoff**

Check sensitivity and specificity tradeoff to find the optimal cutoff point.

7.3.2 Predict on training set at various probability cutoffs <font color="red">[1 Mark]</font>

```python
# Predict on training data by creating columns with different probability cutoffs to explore the impact of cutoff on model performance
numbers = [float(x) / 10 for x in range(10)]
for i in numbers:
    y_train_pred_final[i] = y_train_pred_final["Stayed_Prob"].map(lambda x: 1 if x > i else 0)
y_train_pred_final.head()
```

7.3.3 Plot for accuracy, sensitivity, specificity at different probability cutoffs <font color="red">[2 Marks]</font>

```python
# Create a DataFrame to see the values of accuracy, sensitivity, and specificity at different values of probability cutoffs
rows = []
for i in numbers:
    cm = metrics.confusion_matrix(y_train_pred_final["Stayed"], y_train_pred_final[i])
    acc = (cm[0, 0] + cm[1, 1]) / cm.sum()
    sensi = cm[1, 1] / (cm[1, 0] + cm[1, 1])
    speci = cm[0, 0] / (cm[0, 0] + cm[0, 1])
    rows.append([i, acc, sensi, speci])

cutoff_df = pd.DataFrame(rows, columns=["probability", "accuracy", "sensitivity", "specificity"])
cutoff_df.round(4)
```

**Observation:** As the cutoff rises, sensitivity falls and specificity rises. At 0.5 sensitivity (0.7535) is still above specificity (0.7204), and at 0.6 the order has flipped (0.6474 against 0.8219), so the two curves cross between 0.5 and 0.6. Accuracy is highest at 0.5 (0.7377) and changes little between 0.4 and 0.6. The next cell checks that range in steps of 0.01.

```python
# Check cutoffs between 0.50 and 0.60 in steps of 0.01 to see where sensitivity and specificity meet
rows = []
for i in [x / 100 for x in range(50, 61)]:
    pred = y_train_pred_final["Stayed_Prob"].map(lambda x: 1 if x > i else 0)
    cm = metrics.confusion_matrix(y_train_pred_final["Stayed"], pred)
    acc = (cm[0, 0] + cm[1, 1]) / cm.sum()
    sensi = cm[1, 1] / (cm[1, 0] + cm[1, 1])
    speci = cm[0, 0] / (cm[0, 0] + cm[0, 1])
    rows.append([i, acc, sensi, speci])

fine_df = pd.DataFrame(rows, columns=["probability", "accuracy", "sensitivity", "specificity"])
optimal_cutoff = float(fine_df.loc[(fine_df["sensitivity"] - fine_df["specificity"]).abs().idxmin(), "probability"])
print(fine_df.round(4).to_string(index=False))
print("\nOptimal cutoff:", optimal_cutoff)
```

```python
# Plot accuracy, sensitivity, and specificity at different values of probability cutoffs
plt.figure(figsize=(8, 5))
plt.plot(cutoff_df["probability"], cutoff_df["accuracy"], marker="o", linewidth=2, color=blue, label="Accuracy")
plt.plot(cutoff_df["probability"], cutoff_df["sensitivity"], marker="o", linewidth=2, color=orange, label="Sensitivity")
plt.plot(cutoff_df["probability"], cutoff_df["specificity"], marker="o", linewidth=2, color=aqua, label="Specificity")
plt.axvline(optimal_cutoff, color=grey, linestyle="--", linewidth=1)
plt.text(optimal_cutoff - 0.015, 0.06, f"Chosen cutoff = {optimal_cutoff}", color=grey, ha="right")
plt.xlabel("Probability cutoff")
plt.ylabel("Score")
plt.ylim(0, 1.03)
plt.title("Accuracy, sensitivity and specificity at each cutoff (training data)")
plt.legend(loc="center left", bbox_to_anchor=(1.01, 0.5))
plt.savefig("images/cutoff-tradeoff.png", dpi=150, bbox_inches="tight")
plt.show()
```

**Observations:**
- Sensitivity and specificity are closest at 0.52 (0.7395 and 0.7341), so 0.52 is the optimal cutoff. Accuracy there is 0.7370, almost the same as at 0.5.
- Moving from 0.5 to 0.52 trades a little sensitivity for a little more specificity, which balances the errors between the two classes.
- This choice treats the two kinds of mistake as equally costly: predicting that someone will stay when they leave, and predicting that someone will leave when they stay. If HR could put a cost on each, the cutoff should move to reflect it.

## Conclusion



After dropping 3,975 rows with missing values and 77 duplicate records, fixing the garbled apostrophes in Education Level and removing `Employee ID`, the model was trained on 49,390 employees and checked on 21,168 it had not seen.

The final model is a logistic regression on the 15 features chosen by RFE. All of them are significant (largest p-value 8.8e-27) and none is collinear with the others (largest VIF 1.84), so no feature had to be removed. With a cutoff of 0.52, where sensitivity and specificity meet on the training data, it scores 0.7394 accuracy, 0.7397 sensitivity, 0.7391 specificity and 0.7553 precision on the validation set, with an AUC of 0.8249. These match the training figures closely, so the model generalises well to new employees.

**What drives retention in this data**, holding the other features fixed:
1. Seniority matters most. Senior staff have about 12 times the odds of staying of entry-level staff, and mid-level staff about 2.6 times.
2. Remote work is the clearest policy lever. Remote workers have about 5.6 times the odds of staying.
3. Work-life balance comes next. Employees who rate it Fair or Poor have roughly a third of the odds of staying of those who rate it Good or Excellent. Overtime lowers the odds too (0.72).
4. Poor or Fair company reputation and Low or Below Average performance ratings also lower the odds (0.48 to 0.72).
5. Gender, marital status and education also appear: men, married or divorced employees and PhD holders are more likely to stay. These help the model predict, but gender and marital status are sensitive characteristics, protected by employment law in many countries, so the model should not be used to treat people differently because of them. They are better read as a prompt to ask why the experience of work differs between these groups.

**Suggestions for HR:**
- Review who can work remotely, since remote workers stay far more often.
- Keep overtime in check, and follow up early on poor work-life balance scores.
- Focus retention effort on entry-level staff, for example through clearer progression to mid-level roles, because they leave the most.
- Use the model as a ranked at-risk list. In the validation data, 72.29% of the employees it flagged as likely to leave did leave.

**Limitations:**
- The data looks synthetic. 40.0% of employees would have joined before 18, Monthly Income contains placeholder values around 50,000, and Very High job satisfaction behaves like Low. The coefficients describe this dataset and may not carry over to a real workforce.
- RFE kept only dummy columns. Because a dummy's coefficient and a standardised column's coefficient are not on the same scale, numerical features such as Distance from Home (rank 3) may have been dropped too early. Testing them in the model, or using a selection method that is not based on coefficient size, is the obvious next step.
- Dropping the rows with missing values removed slightly more stayers than leavers, since employees with a missing tenure stayed more often (62.7% against 52.2%).
- The 0.52 cutoff assumes that both kinds of mistake cost the same.
- The model shows association, not cause. Remote work, for example, may be offered more often to people who were already likely to stay.
